### 8. Find the maximum and minimum values of the output of 7.

### 9. Find the maximum and minimum index values of the output of 7.

````python
# 1. Trouver d'abord l'index à plat global (flat index)
flat_argmax = tensor_c_gpu.argmax()
flat_argmin = tensor_c_gpu.argmin()

# 2. Convertir cet index en coordonnées 2D (ligne, colonne)
coords_max = torch.unravel_index(flat_argmax, tensor_c_gpu.shape)
coords_min = torch.unravel_index(flat_argmin, tensor_c_gpu.shape)

print(f"Index max (ligne, colonne) : {coords_max}")
print(f"Index min (ligne, colonne) : {coords_min}")
````
